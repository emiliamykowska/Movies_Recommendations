```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer, MinMaxScaler
```

```python
movies_df = pd.read_csv("ml-latest-small/movies.csv")
ratings_df = pd.read_csv("ml-latest-small/ratings.csv")
tags_df = pd.read_csv("ml-latest-small/tags.csv")
```

```python
movies_df["production_year"] = movies_df.title.str.extract(r"\((\d{4})\)", expand=False).astype(float) #extracting years from titles

movies_df["title"] = movies_df.title.str.replace(pat=r"\(\d{4}\)", repl="", regex=True) #deleteing years of production from titles

movies_df["genres"] = movies_df.genres.str.split("|")

movies_df.head()

print(movies_df["production_year"].min(), movies_df["production_year"].max()) #range of production years
```

```python
mlb = MultiLabelBinarizer()

genres_encoded = mlb.fit_transform(movies_df.genres)

genres_df = pd.DataFrame(genres_encoded, columns=mlb.classes_, index=movies_df.index)

movies_df = pd.concat([movies_df, genres_df], axis=1) #join movies_df and genres_df along columns

# movies_df["genres_encoded"] = genres_encoded.tolist()

# ohe = OneHotEncoder(sparse_output=False)

# years_encoded = ohe.fit_transform(movies_df.production_year.values.reshape(-1, 1))
# movies_df["years_encoded"] = years_encoded.tolist()

# print(movies_df["years_encoded"].iloc[0])
# movies_df.head()

mms = MinMaxScaler(feature_range=(-1, 1))

movies_df["years_encoded"] = mms.fit_transform(movies_df[["production_year"]]) #[[]] to get 2d array not 1d

movies_df.head()
```
